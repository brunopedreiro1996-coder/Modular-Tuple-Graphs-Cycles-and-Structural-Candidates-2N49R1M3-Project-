### Função para Decomposição `N = a * 2^b + c * 3^d`

Com base na descrição fornecida na seção "Decomposição Decimal-Modular por Sufixo Divisível por 3", esta função calcula os parâmetros `a, b, c, d` para um dado número `N`.

```python
def v_k(n, k):
    """Calcula o expoente de k na fatoração de n (i.e., a maior potência de k que divide n)."""
    if n == 0:
        return 0 # Ou raise ValueError, dependendo do contexto
    count = 0
    while n > 0 and n % k == 0:
        count += 1
        n //= k
    return count

def decompose_N(N):
    """Decompõe N na forma a * 2^b + c * 3^d seguindo as regras do documento.

    Args:
        N (int): O número inteiro positivo a ser decomposto.

    Returns:
        tuple: Uma tupla (a, b, c, d) ou None se a decomposição não for encontrada.
    """
    if N <= 0:
        raise ValueError("N deve ser um número inteiro positivo.")

    str_N = str(N)
    len_N = len(str_N)

    # 1. Escolha do Sufixo (S): menor índice L >= 1 para o qual S é divisível por 3.
    selected_L = -1
    for L in range(1, len_N + 1):
        S_candidate = N % (10**L)
        if S_candidate > 0 and S_candidate % 3 == 0: # S > 0 para evitar S=0 caso N seja divisível por 10^L
            selected_L = L
            S = S_candidate
            break

    if selected_L == -1: # Se nenhum sufixo > 0 for divisível por 3
        if N % 3 == 0: # Se N inteiro for divisível por 3, tratamos como (0,0,c,d)
            d = v_k(N, 3)
            c = N // (3**d)
            # print(f"Aviso: Para N={N}, nenhum sufixo decimal é divisível por 3, mas N é. Usando a=0, b=0, c={c}, d={d}.")
            return 0, 0, c, d
        else:
            # print(f"Aviso: Para N={N}, nenhum sufixo decimal ou o próprio N é divisível por 3. Não é possível aplicar a regra de decomposição do sufixo. Retornando None.")
            return None

    P = N // (10**selected_L)
    # S já está definido acima

    # 2. Normalização da Parte Prefixal: P * 10^L = a * 2^b
    prefix_part = P * (10**selected_L)
    if prefix_part == 0:
        b = 0
        a = 0
    else:
        b = v_k(prefix_part, 2)
        a = prefix_part // (2**b)

    # 3. Normalização da Parte Sufixal: S = c * 3^d
    d = v_k(S, 3)
    c = S // (3**d)

    return a, b, c, d
```

```python
df_semiprimes_large = None # Clear previous large dataframe to save memory if necessary

import math
import pandas as pd

MAX_N = 3 * 10**7

# Step 1: Implementar o Crivo de Eratóstenes para gerar o array SPF (Smallest Prime Factor)
def generate_spf_sieve(limit):
    spf = [0] * (limit + 1)
    for i in range(2, limit + 1):
        spf[i] = i
    for i in range(2, int(math.sqrt(limit)) + 1):
        if spf[i] == i: # i é primo
            for j in range(i * i, limit + 1, i):
                if spf[j] == j: # Atualiza apenas se não foi marcado por um primo menor
                    spf[j] = i
    return spf

print(f"Gerando array SPF até {MAX_N}...")
square_root_limit = int(math.sqrt(MAX_N)) + 1
spf_array = generate_spf_sieve(MAX_N)
print("Array SPF gerado com sucesso.")

# Step 2: Gerar semiprimos filtrados usando o array SPF
filtered_semiprimes_data = []

print(f"Encontrando semiprimos filtrados até {MAX_N} (sem filtro N % 7 != 0)...")
for N_val in range(2, MAX_N + 1):
    if spf_array[N_val] != N_val: # N_val é composto
        p = spf_array[N_val]
        q = N_val // p

        # Verifica se q é primo e se p*q é um semiprimo (p <= q)
        # e se p > 5 conforme o documento para iteração de ciclos
        if spf_array[q] == q and p <= q and p > 5:
            # Aplicar os filtros gerais para N
            # REMOVIDO: (N_val % 7 != 0) para incluir números como 119
            if (N_val % 6 == 1 or N_val % 6 == 5) and (N_val % 5 != 0):
                filtered_semiprimes_data.append({'N': N_val, 'p': p, 'q': q})

df_semiprimes_large = pd.DataFrame(filtered_semiprimes_data)
df_semiprimes_large = df_semiprimes_large.sort_values(by='N').reset_index(drop=True)
print(f"Total de {len(df_semiprimes_large)} semiprimos filtrados encontrados até {MAX_N}.")
display(df_semiprimes_large.head())
```

```python
#@title 3. Aplicação da Decomposição `a*2^b+c*3^d`
# Esta célula aplica a função `decompose_N` a todos os semiprimos gerados
# e armazena os resultados em um DataFrame `df_full_analysis`.

decomposition_results_large = []
print(f"Aplicando decompose_N a {len(df_semiprimes_large)} semiprimos...")

# Contadores para avisos
none_decompositions_count = 0

for index, row in df_semiprimes_large.iterrows():
    N = row['N']
    p = row['p']
    q = row['q']
    result = decompose_N(N)
    if result:
        a, b, c, d = result
        decomposition_results_large.append({'N': N, 'p': p, 'q': q, 'a': a, 'b': b, 'c': c, 'd': d})
    else:
        none_decompositions_count += 1

df_full_analysis = pd.DataFrame(decomposition_results_large)
print(f"Foram decompostos {len(df_full_analysis)} semiprimos com os parâmetros a, b, c, d.")
if none_decompositions_count > 0:
    print(f"Aviso: {none_decompositions_count} semiprimos não puderam ser decompostos de acordo com as regras (nenhum sufixo decimal ou N é divisível por 3).")
display(df_full_analysis.head())
```

```python
df_full_analysis['x'] = df_full_analysis['N'] - (df_full_analysis['q'] - df_full_analysis['p'])
df_full_analysis['y'] = df_full_analysis['N'] + (df_full_analysis['q'] - df_full_analysis['p'])
df_full_analysis['w'] = df_full_analysis['N'] - (df_full_analysis['q'] + df_full_analysis['p'])
df_full_analysis['z'] = df_full_analysis['N'] + (df_full_analysis['q'] + df_full_analysis['p'])

# Calcular os valores módulo 6
df_full_analysis['N_mod_6'] = df_full_analysis['N'] % 6
df_full_analysis['w_mod_6'] = df_full_analysis['w'] % 6
df_full_analysis['x_mod_6'] = df_full_analysis['x'] % 6
df_full_analysis['y_mod_6'] = df_full_analysis['y'] % 6
df_full_analysis['z_mod_6'] = df_full_analysis['z'] % 6

# Re-indexar o dataframe para facilitar lookups futuros
df_full_analysis_indexed = df_full_analysis.set_index('N')

display(df_full_analysis.head())
```

```python
df_full_analysis_indexed = df_full_analysis.set_index('N')

def get_all_valid_next_N_primes(current_N):
    """Calcula todos os próximos candidatos (w,x,y,z) e verifica as condições de semiprimo p > 5.
    Retorna uma lista de N' válidos.
    """
    valid_next_Ns = []
    # Usar df_full_analysis_indexed para lookup eficiente
    if current_N not in df_full_analysis_indexed.index:
        return []

    row = df_full_analysis_indexed.loc[current_N]
    candidates = [row['w'], row['x'], row['y'], row['z']]

    for next_N_candidate in candidates:
        # Check if candidate is integer and if it's present in our valid semiprime dataset
        # next_N_candidate could be float if original w,x,y,z were. Ensure it's an int.
        if pd.notna(next_N_candidate) and int(next_N_candidate) in df_full_analysis_indexed.index:
            # A condição p' > 5 já é aplicada durante a geração de df_semiprimes_large
            # e df_full_analysis construction, então se está no DF indexado, é válido.
            valid_next_Ns.append(int(next_N_candidate))

    return list(set(valid_next_Ns)) # Return unique valid next Ns

print("Função 'get_all_valid_next_N_primes' definida para identificar todos os sucessores válidos (w,x,y,z).")
```

### Reconstrução do Grafo com Múltiplas Transições e Detecção de Ciclos

Agora que a função `get_all_valid_next_N_primes` está definida para encontrar todos os sucessores válidos (`w,x,y,z`), vamos reconstruir o grafo e executar a detecção de ciclos. Isso deve permitir a identificação de ciclos-espelho e ciclos genuínos que dependem de qualquer uma dessas transições.

```python
import networkx as nx

# Criar um novo grafo direcionado
G_multi_edge = nx.DiGraph()

# Adicionar nós e arestas ao grafo usando a nova função
print("Construindo grafo com transições múltiplas (w,x,y,z)...")
for N_val in df_full_analysis['N'].unique():
    if N_val in df_full_analysis_indexed.index: # Verificar se N_val está no índice (é um semiprimo decomposto)
        next_N_candidates = get_all_valid_next_N_primes(N_val)
        for next_N in next_N_candidates:
            G_multi_edge.add_edge(N_val, next_N)

print(f"Grafo construído com {G_multi_edge.number_of_nodes()} nós e {G_multi_edge.number_of_edges()} arestas.")

# Re-usar a função find_cycles_dfs para buscar ciclos no novo grafo
# Esta função encontra ciclos usando DFS
def find_cycles_dfs(graph):
    all_cycles = []
    path = []
    visited = set()
    recursion_stack = set()

    def dfs(u):
        visited.add(u)
        recursion_stack.add(u)
        path.append(u)

        for v in graph.neighbors(u):
            if v not in visited:
                dfs(v)
            elif v in recursion_stack:
                # Ciclo encontrado
                cycle_start_index = path.index(v)
                cycle = path[cycle_start_index:]
                all_cycles.append(tuple(cycle))

        path.pop()
        recursion_stack.remove(u)

    for node in graph.nodes():
        if node not in visited:
            dfs(node)

    return all_cycles

print("Buscando ciclos no novo grafo...")
cycles_multi_edge = find_cycles_dfs(G_multi_edge)
print(f"Encontrados {len(cycles_multi_edge)} ciclos totais no grafo de múltiplas transições.")

# Classificar ciclos novamente
ciclos_espelho_multi = []
ciclos_genuinos_multi = []

for cycle in cycles_multi_edge:
    # A lógica para ciclos-espelho (N1 -> N2 -> N1) precisa de cuidado extra
    # para garantir que as arestas existam em ambas as direções diretamente.
    if len(cycle) == 2:
        n1, n2 = cycle[0], cycle[1]
        if G_multi_edge.has_edge(n1, n2) and G_multi_edge.has_edge(n2, n1):
            # Adicionar apenas uma vez para evitar duplicatas (e.g., (119,143) e (143,119))
            # e garantir que a ordem não importa para a unicidade do ciclo
            if tuple(sorted(cycle)) not in [tuple(sorted(c)) for c in ciclos_espelho_multi]:
                ciclos_espelho_multi.append(cycle)
    elif len(cycle) >= 3:
        # Para ciclos genuínos, a ordem importa, mas queremos ciclos únicos.
        # Converter para um tuple e adicionar a um set pode ajudar a evitar duplicatas.
        if tuple(cycle) not in ciclos_genuinos_multi:
            ciclos_genuinos_multi.append(cycle)

print(f"Total de ciclos-espelho (N1 <-> N2): {len(ciclos_espelho_multi)}")
print(f"Total de ciclos genuínos (comprimento >= 3): {len(ciclos_genuinos_multi)}")

print("\nPrimeiros 5 ciclos-espelho:")
for i, cycle in enumerate(ciclos_espelho_multi[:5]):
    print(f"  {cycle}")

print("\nPrimeiros 5 ciclos genuínos:")
for i, cycle in enumerate(ciclos_genuinos_multi[:5]):
    print(f"  {cycle}")

# Exemplo de verificação do ciclo 119 <-> 143 (deve ser um ciclo-espelho)
found_119_143 = False
for cycle in ciclos_espelho_multi:
    if (119 in cycle and 143 in cycle) and len(cycle) == 2:
        found_119_143 = True
        break

if found_119_143:
    print("\nO ciclo-espelho (119, 143) foi encontrado no novo grafo!")
else:
    print("\nO ciclo-espelho (119, 143) AINDA NÃO foi encontrado no novo grafo. Algo está errado na lógica.")
```

### Construção do Grafo de Iteração: Definição de N' = z(N)

Com base no documento do usuário e no exemplo `119↔143`, onde `z(119)` é igual a `143`, assumimos que o próximo `N'` na iteração é dado por `z(N)`. Precisamos garantir que este `N'` também seja um semiprimo válido (conforme nossos filtros e com `p > 5`).

```python
def get_all_valid_next_N_primes(current_N):
    """Calcula todos os próximos candidatos (w,x,y,z) e verifica as condições de semiprimo p > 5.
    Retorna uma lista de N' válidos.
    """
    valid_next_Ns = []
    # Usar df_full_analysis_indexed para lookup eficiente
    if current_N not in df_full_analysis_indexed.index:
        return []

    row = df_full_analysis_indexed.loc[current_N]
    candidates = [row['w'], row['x'], row['y'], row['z']]

    for next_N_candidate in candidates:
        # Check if candidate is integer and if it's present in our valid semiprime dataset
        # next_N_candidate could be float if original w,x,y,z were. Ensure it's an int.
        if pd.notna(next_N_candidate) and int(next_N_candidate) in df_full_analysis_indexed.index:
            # A condição p' > 5 já é aplicada durante a geração de df_semiprimes_large
            # e df_full_analysis construction, então se está no DF indexado, é válido.
            valid_next_Ns.append(int(next_N_candidate))

    return list(set(valid_next_Ns)) # Return unique valid next Ns

print("Função 'get_all_valid_next_N_primes' definida para identificar todos os sucessores válidos (w,x,y,z).")
```

### Construção do Grafo e Detecção de Ciclos

Agora que temos a coluna `next_N` calculada, podemos construir o grafo de iteração e implementar a lógica para encontrar ciclos. Utilizaremos um algoritmo de busca em profundidade (DFS) para identificar ciclos, diferenciando ciclos-espelho de ciclos genuínos.

```python
import networkx as nx
import pandas as pd

# Criar um novo grafo direcionado
G_multi_edge = nx.DiGraph()

# Adicionar nós e arestas ao grafo usando a nova função
print("Construindo grafo com transições múltiplas (w,x,y,z)...")
for N_val in df_full_analysis['N'].unique():
    # Verificar se N_val ainda está no índice df_full_analysis_indexed
    # (para casos em que o N original foi filtrado da decomposição)
    if N_val in df_full_analysis_indexed.index:
        next_N_candidates = get_all_valid_next_N_primes(N_val)
        for next_N in next_N_candidates:
            G_multi_edge.add_edge(N_val, next_N)

print(f"Grafo construído com {G_multi_edge.number_of_nodes()} nós e {G_multi_edge.number_of_edges()} arestas.")

# --- DEBUG START ---
print("\n--- Debugging 119-143 cycle ---")
if G_multi_edge.has_node(119):
    print(f"Node 119 exists in graph.")
    print(f"Neighbors of 119: {list(G_multi_edge.neighbors(119))}")
else:
    print("Node 119 does NOT exist in graph.")

if G_multi_edge.has_node(143):
    print(f"Node 143 exists in graph.")
    print(f"Neighbors of 143: {list(G_multi_edge.neighbors(143))}")
else:
    print("Node 143 does NOT exist in graph.")

if G_multi_edge.has_edge(119, 143):
    print("Edge (119, 143) exists.")
else:
    print("Edge (119, 143) does NOT exist.")

if G_multi_edge.has_edge(143, 119):
    print("Edge (143, 119) exists.")
else:
    print("Edge (143, 119) does NOT exist.")
print("--- Debugging 119-143 cycle END ---\n")
# --- DEBUG END ---

# Re-usar a função find_cycles_dfs para buscar ciclos no novo grafo
# Esta função encontra ciclos usando DFS
def find_cycles_dfs(graph):
    all_cycles = []
    path = []
    visited_nodes = set() # Renomeado para evitar conflito com 'visited' interno do DFS
    recursion_stack = set()

    def dfs(u):
        visited_nodes.add(u)
        recursion_stack.add(u)
        path.append(u)

        for v in graph.neighbors(u):
            if v not in visited_nodes:
                dfs(v)
            elif v in recursion_stack:
                # Ciclo encontrado
                cycle_start_index = path.index(v)
                cycle = path[cycle_start_index:]
                all_cycles.append(tuple(cycle))

        path.pop()
        recursion_stack.remove(u)

    for node in graph.nodes():
        if node not in visited_nodes:
            dfs(node)

    return all_cycles

print("Buscando ciclos no novo grafo...")
cycles_multi_edge = find_cycles_dfs(G_multi_edge)

# Filtrar ciclos duplicados (e.g., [1,2,3] e [2,3,1] são o mesmo ciclo rotacionado)
# E também remover duplicatas de ciclos-espelho se forem adicionados duas vezes ((119,143) e (143,119))
processed_cycles = set()
unique_cycles = []

for cycle in cycles_multi_edge:
    if len(cycle) == 2:
        # Para ciclos-espelho, normalizar a representação para evitar duplicatas (e.g., (119,143) e (143,119))
        normalized_cycle = tuple(sorted(cycle))
    else:
        # Para ciclos genuínos, encontrar a rotação lexicograficamente mínima para representação única
        min_cycle = cycle
        for i in range(len(cycle)):
            rotated_cycle = tuple(cycle[i:] + cycle[:i])
            if rotated_cycle < min_cycle:
                min_cycle = rotated_cycle
        normalized_cycle = min_cycle

    if normalized_cycle not in processed_cycles:
        processed_cycles.add(normalized_cycle)
        unique_cycles.append(cycle)

print(f"Encontrados {len(unique_cycles)} ciclos únicos totais no grafo de múltiplas transições.")

# Classificar ciclos novamente
ciclos_espelho_multi = []
ciclos_genuinos_multi = []

for cycle in unique_cycles:
    if len(cycle) == 2:
        n1, n2 = cycle[0], cycle[1]
        # Confirmar que é um ciclo-espelho (N1 -> N2 e N2 -> N1)
        if G_multi_edge.has_edge(n1, n2) and G_multi_edge.has_edge(n2, n1):
            ciclos_espelho_multi.append(cycle)
    elif len(cycle) >= 3:
        # Para ciclos genuínos, já filtramos rotações para ter um único representante
        ciclos_genuinos_multi.append(cycle)

print(f"Total de ciclos-espelho (N1 <-> N2): {len(ciclos_espelho_multi)}")
print(f"Total de ciclos genuínos (comprimento >= 3): {len(ciclos_genuinos_multi)}")

print("\nPrimeiros 5 ciclos-espelho:")
for i, cycle in enumerate(ciclos_espelho_multi[:5]):
    print(f"  {cycle}")

print("\nPrimeiros 5 ciclos genuínos:")
for i, cycle in enumerate(ciclos_genuinos_multi[:5]):
    print(f"  {cycle}")

# Exemplo de verificação do ciclo 119 <-> 143 (deve ser um ciclo-espelho)
found_119_143 = False
for cycle in ciclos_espelho_multi:
    if (119 in cycle and 143 in cycle) and len(cycle) == 2:
        found_119_143 = True
        break

if found_119_143:
    print("\nO ciclo-espelho (119, 143) foi encontrado no novo grafo!")
else:
    print("\nO ciclo-espelho (119, 143) AINDA NÃO foi encontrado no novo grafo. Verificando as arestas manualmente...")
    # Manual check if 119 -> 143 and 143 -> 119 exist as edges
    if G_multi_edge.has_edge(119, 143) and G_multi_edge.has_edge(143, 119):
        print("  As arestas 119 -> 143 e 143 -> 119 existem no grafo. O ciclo-espelho deveria ter sido detectado. (Possível problema na lógica de classificação de ciclos-espelho)")
    else:
        print("  As arestas 119 -> 143 e 143 -> 119 NÃO existem juntas no grafo. Verificando sucessores...")
        # Use df_full_analysis_indexed for consistent lookup
        if 119 in df_full_analysis_indexed.index:
            z_119 = df_full_analysis_indexed.loc[119]['z']
            print(f"  z(119) = {z_119}")
        else:
            print("  119 não é um nó em df_full_analysis_indexed.")

        if 143 in df_full_analysis_indexed.index:
            z_143 = df_full_analysis_indexed.loc[143]['z']
            print(f"  z(143) = {z_143}")
        else:
            print("  143 não é um nó em df_full_analysis_indexed.")

        if G_multi_edge.has_node(119):
            print(f"  119 sucessores no grafo: {list(G_multi_edge.successors(119))}")
        if G_multi_edge.has_node(143):
            print(f"  143 sucessores no grafo: {list(G_multi_edge.successors(143))}")
```
